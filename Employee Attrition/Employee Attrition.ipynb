```python
db["Attrition"].value_counts()
```

```python
db["Attrition_Flag"] = db["Attrition"].apply(lambda x: 1 if x == "Yes" else 0)
```

```python
Attrition_rate = db["Attrition_Flag"].mean() * 100
print(f"Attrition_rate: {Attrition_rate:.2f}%")
```

```python
db["Attrition"].value_counts().plot(kind = "bar", color = ["steelblue", "salmon"])
plt.title("Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Count")
plt.xticks(rotation = 0)
plt.figure(figsize=(10, 4))
plt.show()
```

```python
db["Department"].value_counts()
```

```python
dept_attrition = db.groupby("Department")["Attrition_Flag"].mean()*100
db1 = dept_attrition.reset_index()
db1.columns = ["Department", "Attrition_rate_%"]
db1.head()
```

```python
sns.barplot(data = db1, x = "Department", y = "Attrition_rate_%", hue = "Department", palette = "Set2", legend = False)
plt.title("Attrition Rate by Department")
plt.xlabel("Departemnt")
plt.ylabel("Attrition_rate_%")
plt.tight_layout()
plt.show()
```
